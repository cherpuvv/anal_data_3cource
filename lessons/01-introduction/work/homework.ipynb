## Подготовка

```python
from pathlib import Path
import sys
import pandas as pd

print("Python:", sys.executable)
print("pandas:", pd.__version__)
print("Рабочая папка:", Path.cwd())
DATA = Path("../data")
assert DATA.is_dir(), "Рядом с папкой ноутбука должна находиться data/"
```

## Основной набор decisions.csv

```python
home_decisions = pd.read_csv(
    DATA / "decisions.csv", dtype={"lei": "string", "county_code": "string"}
)
home_outcomes = home_decisions["action_taken"]
home_n = len(home_decisions)
assert home_n > 0
assert home_outcomes.isna().sum() == 0
assert home_outcomes.isin([1, 2, 3]).sum() == home_n
print(home_decisions.shape)
home_outcomes.value_counts(dropna=False)
```

## H1. Доля отказов

```python
denied_flag = home_outcomes.eq(3)
n_denied = denied_flag.sum()
denial_rate =n_denied / home_n

print(f"Количество отказов: {n_denied}")
print(f"Всего заявок: {home_n}")
print(f"Доля отказов: {denial_rate:.2%}")

assert 0 <= denial_rate <= 1
```

## H2. Одобрение без принятия предложения

```python
not_accepted_flag = home_outcomes.eq(2)
n_not_accepted = not_accepted_flag.sum()
not_accepted_rate = n_not_accepted /home_n
rate_difference_pp = (denial_rate - not_accepted_rate) * 100

print(f"Количество:{n_not_accepted}")
print(f"Доля: {not_accepted_rate:.2%}")
print(f"Разность долей: {rate_difference_pp:.2f} п.п.")

assert 0 <= not_accepted_rate <= 1
```
